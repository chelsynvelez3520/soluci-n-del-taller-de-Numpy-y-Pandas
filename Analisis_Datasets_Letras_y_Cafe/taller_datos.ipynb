**Objetivo:** practicar operaciones fundamentales con arreglos de NumPy y la integración de NumPy con DataFrames de Pandas.

## Preparación: importar librerías

Se utilizan los alias estándar solicitados: `np` para NumPy y `pd` para Pandas.

```python
import numpy as np
import pandas as pd

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

# Ejercicio 1. Operaciones fundamentales con NumPy

## 1. Crear un arreglo unidimensional y consultar su forma y tipo de dato

`shape` indica las dimensiones del arreglo y `dtype` muestra el tipo de dato de sus elementos.

```python
arreglo_1d = np.array([10, 20, 30, 40, 50, 60])

print("Arreglo:", arreglo_1d)
print("Forma (shape):", arreglo_1d.shape)
print("Tipo de dato (dtype):", arreglo_1d.dtype)
```

## 2. Cambiar la forma con `reshape()`

Se reorganizan los seis elementos en una matriz de 2 filas y 3 columnas. `reshape()` no cambia los valores, solo su distribución.

```python
matriz_2d = arreglo_1d.reshape(2, 3)

print("Matriz de 2 dimensiones:")
print(matriz_2d)
print("Nueva forma:", matriz_2d.shape)
```

## 3. Generar matrices aleatorias

- `np.random.rand()` genera números decimales entre 0 y 1.
- `np.random.randint()` genera números enteros dentro del intervalo indicado. El límite superior no se incluye.

```python
np.random.seed(42)  # Permite obtener los mismos resultados al volver a ejecutar la celda.

matriz_decimales = np.random.rand(2, 3)
matriz_enteros = np.random.randint(1, 101, size=(2, 3))

print("Matriz aleatoria con rand():")
print(matriz_decimales)

print("\nMatriz aleatoria de enteros con randint():")
print(matriz_enteros)
```

## 4. Crear matrices estructurales

`zeros()` crea una matriz de ceros, `ones()` una de unos y `eye()` una matriz identidad.

```python
matriz_ceros = np.zeros((3, 3))
matriz_unos = np.ones((2, 4))
matriz_identidad = np.eye(3)

print("Matriz de ceros:")
print(matriz_ceros)

print("\nMatriz de unos:")
print(matriz_unos)

print("\nMatriz identidad:")
print(matriz_identidad)
```

## 5. Transponer una matriz

El atributo `.T` intercambia las filas por las columnas.

```python
matriz_original = np.random.randint(1, 20, size=(2, 3))
matriz_traspuesta = matriz_original.T

print("Matriz original:")
print(matriz_original)
print("Forma original:", matriz_original.shape)

print("\nMatriz traspuesta:")
print(matriz_traspuesta)
print("Forma traspuesta:", matriz_traspuesta.shape)
```

## 6. Concatenar arreglos horizontal y verticalmente

`hstack()` une los arreglos por columnas y `vstack()` los apila por filas. Para que ambas operaciones sean válidas, se utilizan arreglos con dimensiones compatibles.

```python
arreglo_a = np.array([[1, 2], [3, 4]])
arreglo_b = np.array([[5, 6], [7, 8]])

union_horizontal = np.hstack((arreglo_a, arreglo_b))
union_vertical = np.vstack((arreglo_a, arreglo_b))

print("Primer arreglo:")
print(arreglo_a)
print("\nSegundo arreglo:")
print(arreglo_b)

print("\nConcatenación horizontal (hstack):")
print(union_horizontal)

print("\nConcatenación vertical (vstack):")
print(union_vertical)
```

## 7. Operaciones estadísticas y acumulativas

Se calculan la suma, el promedio, la mediana, el producto, la desviación estándar, la varianza, el mínimo, el máximo y la suma acumulativa.

```python
datos = np.array([2, 4, 6, 8, 10])

print("Datos:", datos)
print("Suma:", np.sum(datos))
print("Promedio:", np.mean(datos))
print("Mediana:", np.median(datos))
print("Producto:", np.prod(datos))
print("Desviación estándar:", np.std(datos))
print("Varianza:", np.var(datos))
print("Valor mínimo:", np.min(datos))
print("Valor máximo:", np.max(datos))
print("Suma acumulativa:", np.cumsum(datos))
```

**Nota:** NumPy calcula por defecto la desviación estándar y la varianza de la población (`ddof=0`).

## 8. Operaciones aritméticas elemento a elemento

La suma, resta y multiplicación se realizan entre los elementos que ocupan la misma posición en cada arreglo.

```python
arreglo_x = np.array([10, 20, 30, 40])
arreglo_y = np.array([1, 2, 3, 4])

print("Arreglo X:", arreglo_x)
print("Arreglo Y:", arreglo_y)
print("Suma elemento a elemento:", arreglo_x + arreglo_y)
print("Resta elemento a elemento:", arreglo_x - arreglo_y)
print("Multiplicación elemento a elemento:", arreglo_x * arreglo_y)
```

# Ejercicio 2. Integración y manipulación con Pandas

## 1. Crear un DataFrame a partir de un arreglo de NumPy

Se construye una matriz bidimensional y se convierte en un DataFrame. Los nombres de las columnas se indican explícitamente.

```python
datos_numpy = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
    [15, 25, 35]
])

df = pd.DataFrame(datos_numpy, columns=["A", "B", "C"])

print("DataFrame creado desde NumPy:")
display(df)
```

## 2. Crear un segundo DataFrame desde un diccionario de Python

Las claves del diccionario se convierten en nombres de columnas y las listas en sus respectivos valores.

```python
datos_diccionario = {
    "Producto": ["Libro", "Cuaderno", "Lápiz", "Marcador"],
    "Cantidad": [12, 20, 35, 15],
    "Precio": [25000, 8000, 2000, 4500]
}

df_diccionario = pd.DataFrame(datos_diccionario)

print("DataFrame creado desde un diccionario:")
display(df_diccionario)
```

## 3. Convertir un DataFrame a un arreglo de NumPy

Se muestran las dos formas solicitadas: `.to_numpy()` y `.values`. Ambas permiten obtener los datos del DataFrame como un arreglo de NumPy.

```python
arreglo_con_to_numpy = df.to_numpy()
arreglo_con_values = df.values

print("Conversión con to_numpy():")
print(arreglo_con_to_numpy)
print("Tipo:", type(arreglo_con_to_numpy))

print("\nConversión con values:")
print(arreglo_con_values)
print("Tipo:", type(arreglo_con_values))
```

## 4. Calcular promedios con `np.mean()`

- `axis=0`: calcula el promedio de cada columna.
- `axis=1`: calcula el promedio de cada fila.

Se aplica `np.mean()` directamente sobre el DataFrame.

```python
promedio_columnas_numpy = np.mean(df, axis=0)
promedio_filas_numpy = np.mean(df, axis=1)

print("Promedio de cada columna (axis=0):")
print(promedio_columnas_numpy)

print("\nPromedio de cada fila (axis=1):")
print(promedio_filas_numpy)
```

## 5. Calcular promedios con los métodos de Pandas

Se repiten los cálculos utilizando el método nativo `DataFrame.mean()`.

```python
promedio_columnas_pandas = df.mean(axis=0)
promedio_filas_pandas = df.mean(axis=1)

print("Promedio de cada columna con df.mean(axis=0):")
print(promedio_columnas_pandas)

print("\nPromedio de cada fila con df.mean(axis=1):")
print(promedio_filas_pandas)
```

## 6. Comparación de los resultados

Los resultados obtenidos con NumPy y Pandas deben coincidir para estos datos numéricos.

```python
print("¿Coinciden los promedios por columna?",
      np.allclose(promedio_columnas_numpy.to_numpy(), promedio_columnas_pandas.to_numpy()))

print("¿Coinciden los promedios por fila?",
      np.allclose(promedio_filas_numpy.to_numpy(), promedio_filas_pandas.to_numpy()))
```
